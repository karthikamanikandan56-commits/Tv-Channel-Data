```python
numeric_columns = [
    "Avg_Viewers",
    "Avg_Rating",
    "Subscription_Fee",
    "Years_Active",
    "Total_Likes",
    "YouTube_Views",
    "Social_Media_Followers",
    "Monthly_Reach"
]

for column in numeric_columns:

    converted = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    invalid = converted.isna() & df[column].notna()

    print(
        column,
        "text/non-numeric values:",
        invalid.sum()
    )
```

```python
#Data cleaning
df["Channel_Name"] = (
    df["Channel_Name"]
    .astype(str)
    .str.strip()
    .str.title()
)

print(df["Channel_Name"].unique())
```

```python
df = df.replace(
    ["NA", "N/A", "na", "n/a",
     "Unknown", "unknown", ""],
    np.nan
)
numeric_columns = df.select_dtypes(
    include=np.number
).columns

for column in numeric_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )
    numeric_columns = df.select_dtypes(
    include=np.number
).columns

for column in numeric_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )
```

```python
df = df.drop_duplicates()

print("Duplicate Records:", df.duplicated().sum())
```

```python
df.loc[
    (df["Avg_Rating"] < 0) |
    (df["Avg_Rating"] > 5),
    "Avg_Rating"
] = np.nan

df["Avg_Rating"] = df["Avg_Rating"].fillna(
    df["Avg_Rating"].median()
)

print(df["Avg_Rating"].describe())
```

```python
df.loc[
    df["Subscription_Fee"] < 0,
    "Subscription_Fee"
] = np.nan

df["Subscription_Fee"] = df[
    "Subscription_Fee"
].fillna(
    df["Subscription_Fee"].median()
)

print(df["Subscription_Fee"].describe())
```

```python
df.loc[
    (df["Years_Active"] < 0) |
    (df["Years_Active"] > 100),
    "Years_Active"
] = np.nan

df["Years_Active"] = df[
    "Years_Active"
].fillna(
    df["Years_Active"].median()
)

print(df["Years_Active"].describe())
```

```python
df.loc[
    (df["Years_Active"] < 0) |
    (df["Years_Active"] > 100),
    "Years_Active"
] = np.nan

df["Years_Active"] = df[
    "Years_Active"
].fillna(
    df["Years_Active"].median()
)

print(df["Years_Active"].describe())
```
